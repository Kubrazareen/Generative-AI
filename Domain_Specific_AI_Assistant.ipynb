### 📘 Project Overview
This notebook demonstrates how to build a **Domain-Specific AI Code Assistant** using a **Large Language Model (LLM) API**.  
The assistant can:
- Explain, fix, or complete code snippets.
- Adapt its responses based on a chosen domain (e.g., Web Development, Data Science).
- Provide intelligent feedback and code suggestions using an API (no training required).

**Objective:** Learn how to integrate an LLM API into a code editor or environment to build a simple intelligent assistant similar to GitHub Copilot.

---


### 🧩 Step 1: Setup & Imports

```

import os
import requests
from dotenv import load_dotenv

# Load environment variables (API keys, URLs)
load_dotenv()

# If you don't have a .env file, set your API key manually (not recommended for public use)
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY", "your_api_key_here")
LLM_API_URL = os.getenv("LLM_API_URL", "https://api.openai.com/v1/chat/completions")
MODEL_NAME = os.getenv("MODEL_NAME", "gpt-4o-mini")

print("✅ Environment setup complete.")
```

### 🧠 Step 2: Build the Structured Prompt

```

def build_prompt(code, language, task, domain="General"):
    """Builds a structured prompt for the LLM to analyze code."""
    system_prompt = f"You are an expert {language} developer for domain: {domain}. Respond accurately and concisely."
    
    user_prompt = f"""
    Task: {task}
    Code:
    {code}

    Please provide:
    1. Explanation of what the code does.
    2. Potential issues or improvements.
    3. Final corrected/optimized code (if needed).
    """
    
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
```

### 🌐 Step 3: Send Request to LLM API

```

def query_llm(messages):
    headers = {
        "Authorization": f"Bearer {OPENAI_API_KEY}",
        "Content-Type": "application/json"
    }

    payload = {
        "model": MODEL_NAME,
        "messages": messages,
        "max_tokens": 800,
        "temperature": 0.3
    }

    try:
        response = requests.post(LLM_API_URL, headers=headers, json=payload, timeout=30)
        response.raise_for_status()
        data = response.json()
        return data["choices"][0]["message"]["content"]
    
    except Exception as e:
        return f"❌ Error: {e}"
```

### 💬 Step 4: Interactive Domain-Specific Assistant

```

def domain_specific_assistant():
    print("🤖 Domain-Specific AI Code Assistant")
    print("====================================")
    language = input("Enter programming language (e.g., Python, JavaScript): ")
    domain = input("Enter your domain (e.g., Web Development, Data Science): ")
    task = input("Enter task (explain / fix / complete): ")

    print("\nPaste your code below (finish input with ENTER + twice):")
    print("--------------------------------------------------------")
    
    lines = []
    while True:
        line = input()
        if line == "":
            break
        lines.append(line)
    code = "\n".join(lines)

    messages = build_prompt(code, language, task, domain)
    print("\n🧠 Generating response from LLM...\n")
    result = query_llm(messages)
    print("💬 LLM Response:\n")
    print(result)
```

### ▶️ Step 5: Run the Assistant

```
domain_specific_assistant()
```

### 🧪 Step 6: Quick Test (Non-Interactive)

```

test_code = """
def add_numbers(a, b):
    return a + b
"""

messages = build_prompt(test_code, "Python", "explain", "Mathematics Utility")
result = query_llm(messages)
print("🧠 Model Output:\n")
print(result)
```

### 💾 Step 7: Save Output to File

```

# Save LLM response to a text file
with open("llm_output.txt", "w", encoding="utf-8") as f:
    f.write(result)
print("✅ Output saved to llm_output.txt")
```
